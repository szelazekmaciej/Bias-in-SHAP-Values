# Preprocess and SHAP values

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    balanced_accuracy_score,
    accuracy_score,
    cohen_kappa_score,
    f1_score,
    roc_auc_score,
)
import pickle
import shap
import numpy as np

# Lista para almacenar los resultados de cada modelo y métricas adicionales
results = []

# Bucle para cargar, entrenar y evaluar un Random Forest en cada dataset
for j in range(1, 6):
    for i in range(1, 8):
        # Cargar el dataset
        filename = f'list_ir_{j}_subset_{i}.csv'
        df = pd.read_csv(f'Datasets/list_ir_{j}_subset_{i}.csv')

        # Separar la columna 'EDAD' (variable numérica) y las variables categóricas
        X_num = df[['EDAD']]  # Seleccionar solo la columna 'EDAD'
        X_cat = df.drop(columns=['SITUACION', 'EDAD', 'pauta', "ID_PACIENTE"])  # Excluir columnas
        X_cat = pd.get_dummies(X_cat, drop_first=True)  # Realizar one-hot encoding

        # Concatenar las variables numéricas y categóricas codificadas
        X = pd.concat([X_num, X_cat], axis=1)

        # Variable objetivo
        y = df['SITUACION']
        ids = df['ID_PACIENTE']
        
        column_name_mapping = {
            'EDAD': 'AGE',
            'SEXO_M': 'SEX_FEMALE',
            'DM_SI': 'DIABETES',
            'IC_SI': 'HEART_FAILURE',
            'EPOC_SI': 'COPD',
            'HTA_SI': 'HYPERTENSION',
            'DEP_SI': 'DEPRESSION',
            'CI_SI': 'ISCHEMIC_HEART_DISEASE',
            'ACV_SI': 'STROKE',
            'IRC_SI': 'CHRONIC_KIDNEY_DISEASE',
            'CIR_SI': 'CIRRHOSIS',
            'OST_SI': 'OSTEOPOROSIS',
            'ARTROSIS_SI': 'ARTHROSIS',
            'ARTRITIS_SI': 'ARTHRITIS',
            'DEM_SI': 'DEMENTIA',
            'Obesidad_SI': 'OBESITY',
            'Asma_SI': 'ASTHMA',
            'ID_PACIENTE': 'PATIENT_ID',
        }
        X= X.rename(columns=column_name_mapping)
        
        feature_names = X.columns.tolist()
        
        # Entrenar el modelo Random Forest
        rf = RandomForestClassifier(
            n_estimators=100, max_depth=5, class_weight='balanced', random_state=42
        )
        rf.fit(X, y)

        # Realizar predicciones y calcular métricas
        y_pred = rf.predict(X)
        y_pred_proba = rf.predict_proba(X)[:, 1]

        # Mapear las etiquetas a valores numéricos
        label_mapping = {'SI': 1, 'NO': 0}
        y_test_mapped = y.map(label_mapping)
        y_pred_mapped = pd.Series(y_pred).map(label_mapping)

        # Calcular las métricas
        metrics = {
            'Balanced Accuracy': balanced_accuracy_score(y_test_mapped, y_pred_mapped),
            'Accuracy': accuracy_score(y_test_mapped, y_pred_mapped),
            'Cohen\'s Kappa': cohen_kappa_score(y_test_mapped, y_pred_mapped),
            'F1-Score': f1_score(y_test_mapped, y_pred_mapped),
            'ROC-AUC': roc_auc_score(y_test_mapped, y_pred_proba),
        }

        # Guardar las métricas y el modelo
        results.append({'Dataset': filename, **metrics})
        model_filename = f'models/rf_{filename.split(".")[0]}.pkl'
        with open(model_filename, 'wb') as f:
            pickle.dump(rf, f)

        # Guardar SHAP values en un dataset
        explainer = shap.TreeExplainer(rf)
        shap_values = explainer.shap_values(X)

        shap_df = pd.DataFrame(
            shap_values[:,:,0],  # SHAP values para la clase positiva
            columns=feature_names,
        )
        shap_df['PATIENT_ID'] = ids.values
        shap_df['Real_class'] = y.values
        shap_df['Predicted_class'] = y_pred
        shap_df['Model_Output'] = y_pred_proba
        
        shap_filename = f'shap/shap_list_ir_{j}_subset_{i}_class0.csv'
        shap_df.to_csv(shap_filename, index=False)
        
        shap_df = pd.DataFrame(
            shap_values[:,:,1],  # SHAP values para la clase negativa
            columns=feature_names,
        )
        shap_df['PATIENT_ID'] = ids.values
        shap_df['Real_class'] = y.values
        shap_df['Predicted_class'] = y_pred
        shap_df['Model_Output'] = y_pred_proba
        
        shap_filename = f'shap/shap_list_ir_{j}_subset_{i}_class1.csv'
        shap_df.to_csv(shap_filename, index=False)

# Guardar todas las métricas en un archivo CSV
metrics_df = pd.DataFrame(results)
metrics_df.to_csv('models/metrics_summary.csv', index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import glob

# Directorio donde están los archivos .parquet
shap_dir = 'shap_1000_subsets/'
ir_ratios = range(1, 6)  # IR de 1 a 5
results = []

# Procesar cada IR
for ir in ir_ratios:
    # Encontrar todos los archivos para la clase 1 y el IR actual
    files = glob.glob(f"{shap_dir}/shap_list_ir_{ir}_subset_*_class1.parquet")
    all_shap = []

    for file in files:
        # Cargar el archivo y descartar columnas irrelevantes
        df = pd.read_parquet(file)
        df = df.drop(columns=["Real_class", "Predicted_class", "Model_Output"])
        all_shap.append(df)

    # Concatenar todos los SHAP values de los subsets
    shap_combined = pd.concat(all_shap)

    # Calcular la media de SHAP values por paciente
    shap_mean = shap_combined.groupby("PATIENT_ID").mean()

    # Agregar la columna del IR para identificarlo en los gráficos
    shap_mean["IR"] = ir
    results.append(shap_mean)

# Combinar resultados de todos los IR
final_df = pd.concat(results)

# Convertir a formato largo para graficar con seaborn
long_df = final_df.reset_index().melt(id_vars=["PATIENT_ID", "IR"], var_name="Variable", value_name="SHAP_Value")

# === Gráficos ===
# Configuración de estilo
sns.set(style="whitegrid")

# KDE plot para cada variable
for variable in long_df["Variable"].unique():
    plt.figure(figsize=(10, 6))
    sns.kdeplot(data=long_df[long_df["Variable"] == variable], x="SHAP_Value", hue="IR", fill=True, palette="Set2")
    plt.title(f"KDE Plot for {variable}")
    plt.xlabel("SHAP Value")
    plt.ylabel("Density")
    plt.legend(title="IR")
    plt.tight_layout()
    plt.savefig(f"KDE_{variable}.png")  # Guardar el gráfico
    plt.show()

# Boxplot para cada variable
for variable in long_df["Variable"].unique():
    plt.figure(figsize=(10, 6))
    sns.boxplot(data=long_df[long_df["Variable"] == variable], x="IR", y="SHAP_Value", palette="Set2")
    plt.title(f"Boxplot for {variable}")
    plt.xlabel("Imbalanced Ratio (IR)")
    plt.ylabel("SHAP Value")
    plt.tight_layout()
    plt.savefig(f"Boxplot_{variable}.png")  # Guardar el gráfico
    plt.show()
```

```python
# Convertir IR a cadenas con prefijo "IR"
long_df["IR"] = long_df["IR"].astype(str)

# === Gráficos ===
sns.set(style="whitegrid")

# === KDE Plot con ajuste de leyenda ===
for variable in long_df["Variable"].unique():
    plt.figure(figsize=(10, 6))
    ax = sns.kdeplot(
        data=long_df[long_df["Variable"] == variable],
        x="SHAP_Value",
        hue="IR",
        fill=True,
        palette="Set2"
    )
    plt.title(f"KDE Plot for {variable}")
    plt.xlabel("SHAP Value")
    plt.ylabel("Density")
    sns.move_legend(ax, "best", title="Imbalanced Ratio (IR)")
    plt.tight_layout()
    plt.savefig(f"KDE_{variable}.png")  # Guardar el gráfico
    plt.show()


# Boxplot para cada variable
for variable in long_df["Variable"].unique():
    plt.figure(figsize=(10, 6))
    sns.boxplot(data=long_df[long_df["Variable"] == variable], x="IR", y="SHAP_Value", palette="Set2")
    plt.title(f"Boxplot for {variable}")
    plt.xlabel("Imbalanced Ratio (IR)")
    plt.ylabel("SHAP Value")
    plt.tight_layout()
    plt.legend()
    plt.savefig(f"Boxplot_{variable}.png")  # Guardar el gráfico
    plt.show()
#RUG
```

# Medias de los valores absolutos de los SHAP por pacientes
    

```python
import pandas as pd
import glob
import os
import numpy as np
from scipy.stats import ttest_rel

# Carpeta donde están los archivos .parquet
input_folder = "shap_1000_subsets"

# Listar todos los archivos .parquet para clase 1
parquet_files = glob.glob(os.path.join(input_folder, "*class1.parquet"))

# Cargar y combinar los archivos .parquet para cada IR
shap_data = pd.DataFrame()
for file in parquet_files:
    # Extraer el IR desde el nombre del archivo
    ir = os.path.basename(file).split("_")[3]
    temp_df = pd.read_parquet(file)
    temp_df["IR"] = ir  # Añadir la columna IR al DataFrame
    shap_data = pd.concat([shap_data, temp_df], ignore_index=True)

# Filtrar columnas que no son SHAP values
excluded_columns = ["PATIENT_ID", "Real_class", "Predicted_class", "Model_Output", "IR"]
shap_columns = [col for col in shap_data.columns if col not in excluded_columns]

# Calcular la media de los valores absolutos de los SHAP values por paciente y por IR
shap_abs_mean = (
    shap_data.groupby(["PATIENT_ID", "IR"])[shap_columns]
    .apply(lambda df: df.abs().mean())
    .reset_index()
)

# Transformar los datos a un formato wide para análisis pareado
wide_data = shap_abs_mean.pivot(index="PATIENT_ID", columns="IR", values=shap_columns)
```

```python

# Realizar análisis estadístico pareado entre IRs
statistical_results = {}
irs = sorted(shap_abs_mean["IR"].unique())  # Obtener lista de IRs ordenada
for i in range(len(irs) - 1):
    for j in range(i + 1, len(irs)):
        ir1, ir2 = irs[i], irs[j]
        paired_results = {}
        for variable in shap_columns:
            t_stat, p_value = ttest_rel(wide_data[(variable, ir1)], wide_data[(variable, ir2)])
            paired_results[variable] = {"t_stat": t_stat, "p_value": p_value}
        statistical_results[f"{ir1} vs {ir2}"] = paired_results

# Guardar los resultados estadísticos en un archivo CSV
stat_results_df = pd.DataFrame(
    {
        (ir_pair, variable): results
        for ir_pair, variables in statistical_results.items()
        for variable, results in variables.items()
    }
).T
stat_results_df.columns = ["t_stat", "p_value"]
stat_results_df.to_csv("paired_statistical_results.csv")

print("Análisis pareado completado y resultados guardados en paired_statistical_results.csv")

```

## ANOVA

```python
from scipy.stats import shapiro
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

# Test de Shapiro-Wilk por variable e IR
normality_results = {}
for variable in shap_columns:
    for ir in shap_abs_mean["IR"].unique():
        subset = shap_abs_mean[shap_abs_mean["IR"] == ir][variable]
        stat, p_value = shapiro(subset)
        normality_results[(variable, ir)] = {"W-stat": stat, "p-value": p_value}

# Mostrar resultados
for (variable, ir), result in normality_results.items():
    print(f"Normality Test for {variable} in IR {ir}: W-stat = {result['W-stat']}, p-value = {result['p-value']}")

from scipy.stats import levene

# Test de Levene por variable
variance_results = {}
for variable in shap_columns:
    subsets = [shap_abs_mean[shap_abs_mean["IR"] == ir][variable] for ir in shap_abs_mean["IR"].unique()]
    stat, p_value = levene(*subsets)
    variance_results[variable] = {"Levene-stat": stat, "p-value": p_value}

# Mostrar resultados
for variable, result in variance_results.items():
    print(f"Variance Homogeneity Test for {variable}: Levene-stat = {result['Levene-stat']}, p-value = {result['p-value']}")

    
from statsmodels.stats.anova import AnovaRM

# Transformar los datos a formato long para ANOVA
anova_data = shap_abs_mean.melt(id_vars=["PATIENT_ID", "IR"], var_name="Variable", value_name="Mean_Abs_SHAP")

# Realizar el ANOVA para cada variable
anova_results = {}
for variable in shap_columns:
    # Filtrar los datos para la variable actual
    variable_data = anova_data[anova_data["Variable"] == variable]
    
    # Realizar ANOVA de medidas repetidas
    aovrm = AnovaRM(variable_data, depvar="Mean_Abs_SHAP", subject="PATIENT_ID", within=["IR"])
    anova_result = aovrm.fit()
    anova_results[variable] = anova_result.summary()

# Mostrar resultados
for variable, result in anova_results.items():
    print(f"ANOVA Result for Variable: {variable}")
    print(result)
    print("\n")

# Si prefieres guardar los resultados
with open("anova_results.txt", "w") as f:
    for variable, result in anova_results.items():
        f.write(f"ANOVA Result for Variable: {variable}\n")
        f.write(f"{result}\n\n")
```
